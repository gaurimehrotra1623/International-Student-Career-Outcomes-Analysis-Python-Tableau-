```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('student_dataset.csv') 
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df[df["Job_Sector"].isnull()]["Employment_Status"].value_counts() #checking the cause of null job sector 
```

```python
df["Job_Sector"] = df["Job_Sector"].fillna("Not Specified")
```

```python
df.loc[
    df["Job_Sector"].isnull() &
    (df["Employment_Status"] != "Employed"),
    "Job_Sector"
] = "Not Applicable"
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.info()
```

```python
for col in df.select_dtypes(include='object'):
    print(col)
    print(df[col].value_counts().head(10))
    print("------")
```

```python
df["Gender"] = df["Gender"].str.strip().str.title()
```

```python
df["Salary"].describe()
```

```python
df[df["Salary"] == 0]["Employment_Status"].value_counts()
```

```python
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[20,25,30,35,40],
    labels=["20-25","26-30","31-35","36-40"]
)

df["GPA_Category"] = pd.cut(
    df["GPA"],
    bins=[0,2.5,3.0,3.5,4.0,5.0],
    labels=["Low","Average","Good","Very Good","Excellent"]
)
```

```python
df.to_csv("cleaned_dataset.csv", index=False)
```
